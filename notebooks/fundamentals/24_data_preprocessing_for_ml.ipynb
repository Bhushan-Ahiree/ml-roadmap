# Why Data Preprocessing Matters

Machine Learning algorithms expect clean and structured data.

However, real-world datasets often contain:

- Text values
- Different units of measurement
- Missing values
- Duplicate records
- Outliers

Without proper preprocessing, even a good algorithm can produce poor results.

In practice, data preprocessing is one of the most important stages of the Machine Learning workflow.

# Features

A feature is an input variable used by a Machine Learning model.

Example:

Suppose we want to predict the price of a house.

Possible features are:

- Number of Bedrooms
- Square Feet
- Property Type
- Location

The model learns the relationship between these features and the target variable.

# Types of Features

Machine Learning features are broadly divided into two categories:

- Numerical Features
- Categorical Features

Understanding the type of each feature helps us choose the correct preprocessing technique.

# Numerical Features

Numerical features contain measurable numeric values.

Examples:

- Age
- Salary
- Square Feet
- House Price

These features support mathematical operations such as addition, subtraction, and averaging.

```python
import pandas as pd

df = pd.DataFrame(
    {
        "Age": [22, 30, 27],
        "Salary": [350000, 650000, 520000],
    }
)

df
```

# Categorical Features

Categorical features represent groups or labels instead of measurable numbers.

Examples:

- City
- Property Type
- Gender
- Color

Machine Learning algorithms cannot directly understand text values.

These values must first be converted into numerical form.

```python
df = pd.DataFrame(
    {
        "City": ["Pune", "Mumbai", "Nashik"],
        "Property Type": ["Flat", "House", "Villa"],
    }
)

df
```

# Why Can't Models Use Text Directly?

Most Machine Learning algorithms perform mathematical calculations.

For example:

A Linear Regression model cannot calculate using:

"Pune"

or

"Villa"

These values have no mathematical meaning.

Therefore, categorical values must be encoded into numbers before training.

# Types of Categorical Features

Not all categorical features are the same.

They are divided into two categories:

- Nominal Features
- Ordinal Features

Choosing the correct encoding method depends on which type of categorical feature you have.

# Nominal Features

Nominal features represent categories with **no natural order**.

Examples:

- City
- Property Type
- Car Brand
- Blood Group

Example:

Property Type

- Flat
- House
- Villa

These are simply different categories.

There is no ranking between them.

```python
property_types = [
    "Flat",
    "House",
    "Villa",
]

print(property_types)
```

## Why Order Matters

Suppose we encode:

Flat → 0

House → 1

Villa → 2

A Machine Learning model may incorrectly assume:

Villa > House > Flat

But this relationship does not exist.

These numbers represent labels, not rankings.

This is why Label Encoding is usually **not appropriate for nominal features**.

# Ordinal Features

Ordinal features have a meaningful order.

Examples:

- Education Level
- Customer Rating
- T-Shirt Size

Example:

Small

Medium

Large

Here, the order carries information.

Large is greater than Medium, and Medium is greater than Small.

```python
sizes = [
    "Small",
    "Medium",
    "Large",
]

print(sizes)
```

## Label Encoding

Label Encoding assigns an integer value to each category.

Example:

Small → 0

Medium → 1

Large → 2

Since these categories already have an order, Label Encoding preserves useful information.

This makes it suitable for **ordinal features**.

```python
import pandas as pd

df = pd.DataFrame(
    {
        "Size": [
            "Small",
            "Medium",
            "Large",
            "Medium",
        ]
    }
)

mapping = {
    "Small": 0,
    "Medium": 1,
    "Large": 2,
}

df["Encoded Size"] = df["Size"].map(mapping)

df
```

# One-Hot Encoding

One-Hot Encoding creates a new binary column for each category.

Instead of assigning numbers to categories, it represents each category using 0s and 1s.

This avoids introducing an artificial order.

One-Hot Encoding is the preferred choice for **nominal features**.

```python
from sklearn.preprocessing import OneHotEncoder

df = pd.DataFrame(
    {
        "Property Type": [
            "Flat",
            "House",
            "Villa",
            "Flat",
        ]
    }
)

encoder = OneHotEncoder(
    sparse_output=False
)

encoded = encoder.fit_transform(
    df[["Property Type"]]
)

encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out()
)

encoded_df
```
