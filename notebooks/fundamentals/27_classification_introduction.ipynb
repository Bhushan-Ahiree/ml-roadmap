# What is Classification?

Classification is a supervised Machine Learning technique used to predict categories or labels.

Instead of predicting a numerical value, the model predicts which class an observation belongs to.

Examples:

- Spam or Not Spam
- Fraud or Not Fraud
- Disease or No Disease

```python
import pandas as pd

df = pd.DataFrame(
    {
        "Email": [
            "Win a free iPhone!",
            "Meeting at 3 PM",
            "Congratulations! Claim your prize.",
            "Project update attached.",
        ],
        "Label": [
            "Spam",
            "Not Spam",
            "Spam",
            "Not Spam",
        ],
    }
)

df
```

In this example:

The feature is the email content.

The target is the label:

- Spam
- Not Spam

The goal of the model is to learn patterns that distinguish spam emails from legitimate ones.

# Classification vs Regression

Both classification and regression are supervised learning techniques.

The key difference lies in the type of prediction they make.

Regression predicts continuous numerical values.

Classification predicts discrete categories or labels.

```python
comparison = pd.DataFrame(
    {
        "Regression": [
            "House Price",
            "Salary",
            "Temperature",
        ],
        "Classification": [
            "Spam",
            "Fraud",
            "Disease",
        ],
    }
)

comparison
```

Examples:

Regression

- Predict House Price
- Predict Salary
- Predict Temperature

Classification

- Spam Detection
- Fraud Detection
- Disease Diagnosis

The output determines whether the problem is regression or classification.

# Why Does Classification Matter?

Many real-world business problems require making decisions rather than predicting numbers.

Examples include:

- Should a loan be approved?
- Is this transaction fraudulent?
- Is this email spam?
- Will a customer leave the service?

Classification models help automate these decisions using historical data.

# Binary Classification

Binary Classification involves only two possible classes.

Examples:

- Yes / No
- Fraud / Not Fraud
- Spam / Not Spam
- Pass / Fail

The model predicts one of these two outcomes for each observation.

```python
binary_examples = pd.DataFrame(
    {
        "Transaction": [1, 2, 3, 4],
        "Prediction": [
            "Fraud",
            "Not Fraud",
            "Not Fraud",
            "Fraud",
        ],
    }
)

binary_examples
```
