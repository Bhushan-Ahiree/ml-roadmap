There is no single algorithm that performs best on every dataset.

In this notebook, we'll study the most commonly used regression algorithms in scikit-learn and understand when to use each one.

# Why Do We Need Multiple Algorithms?

Imagine solving every problem using only one tool.

A hammer is useful for nails, but not for screws.

Similarly, one Machine Learning algorithm cannot solve every regression problem efficiently.

Different datasets have different characteristics, such as:

- Linear relationships
- Non-linear relationships
- Outliers
- Correlated features
- Noise

Different algorithms handle these situations differently.

# Linear Regression

Linear Regression is the simplest regression algorithm.

It assumes that the relationship between the input features and the target is approximately linear.

It learns a straight line (or a hyperplane for multiple features) that best fits the training data.

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

print(model)
```

## When Should You Use Linear Regression?

Linear Regression works well when:

- The relationship is approximately linear.
- The dataset is relatively clean.
- Model interpretability is important.
- A simple baseline model is needed.

It is often the first algorithm tried for regression problems.

## Advantages

- Simple to understand.
- Fast to train.
- Easy to interpret.
- Good baseline model.

## Limitations

- Assumes a linear relationship.
- Sensitive to outliers.
- May underfit complex datasets.

# Ridge Regression

Ridge Regression is an improved version of Linear Regression.

It keeps the same basic idea of fitting a line but discourages the model from relying too heavily on any single feature.

This helps reduce overfitting and often improves generalization.

```python
from sklearn.linear_model import Ridge

model = Ridge()

print(model)
```

## When Should You Use Ridge Regression?

Use Ridge Regression when:

- Linear Regression performs well but slightly overfits.
- Features are highly correlated.
- You want a more stable linear model.

Ridge usually keeps all features but reduces the influence of less important ones.

# Interview Insight

### Why isn't Linear Regression always the best choice?

Because many real-world datasets have complex, non-linear relationships that a straight line cannot model effectively.

---

### When should Ridge Regression be preferred?

When you want to reduce overfitting while keeping all features in the model.

---

### Is Ridge Regression a different algorithm from Linear Regression?

It is an extension of Linear Regression that adds regularization to improve generalization.

# Lasso Regression

Lasso Regression is another extension of Linear Regression.

Like Ridge Regression, it helps reduce overfitting.

However, Lasso has one unique property:

It can completely remove the influence of less important features by reducing their coefficients to zero.

As a result, Lasso can automatically perform feature selection.

```python
from sklearn.linear_model import Lasso

model = Lasso()

print(model)
```

## When Should You Use Lasso Regression?

Use Lasso Regression when:

- Some features may not be useful.
- You want a simpler model.
- Automatic feature selection is beneficial.

Unlike Ridge Regression, Lasso may completely remove unimportant features from the model.

# Ridge vs Lasso

Although both algorithms reduce overfitting, they behave differently.

Ridge Regression:

- Keeps all features.
- Reduces the influence of less important features.

Lasso Regression:

- Can completely remove less important features.
- Produces a simpler model by performing feature selection.

# ElasticNet

ElasticNet combines the ideas of Ridge Regression and Lasso Regression.

It reduces overfitting while also allowing feature selection.

ElasticNet is particularly useful when:

- Many features are correlated.
- You want a balance between Ridge and Lasso.

```python
from sklearn.linear_model import ElasticNet

model = ElasticNet()

print(model)
```

# Decision Tree Regressor

A Decision Tree predicts values by repeatedly splitting the data into smaller groups based on feature values.

Instead of fitting a straight line, it learns a series of decision rules.

This allows it to capture complex, non-linear relationships.

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(random_state=42)

print(model)
```

## Advantages

- Learns non-linear relationships.
- Easy to understand visually.
- No feature scaling required.

## Limitations

- Can easily overfit the training data.
- Small changes in the data can produce different trees.

# Random Forest Regressor

Random Forest is an ensemble algorithm.

Instead of building one Decision Tree, it builds many trees.

Each tree makes its own prediction.

The final prediction is the average of all tree predictions.

This usually produces better and more stable results than a single Decision Tree.

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(random_state=42)

print(model)
```

## Advantages

- Captures complex patterns.
- Less prone to overfitting than a single Decision Tree.
- No feature scaling required.
- Performs well on many real-world datasets.

## Limitations

- Slower to train than a single Decision Tree.
- Harder to interpret.

# Which Algorithm Should You Choose?

There is no universal best algorithm.

A practical approach is:

1. Start with Linear Regression as a baseline.
2. If overfitting becomes a concern, try Ridge or Lasso.
3. If the relationship is clearly non-linear, try Decision Tree.
4. If a Decision Tree overfits, try Random Forest.
5. Compare models using evaluation metrics instead of assumptions.

Model selection should be driven by performance on unseen data.
